```python
file = open('hostel_bois.txt','r',encoding='utf-8')
content = file.read()
lines = content.split('\n')
len(lines)
lines[-1]
line=lines[3]
print(line)

timestamp,remaining = line.split(' - ',1)
print(timestamp)
print(remaining)

sender,message = remaining.split(': ',1)
print(sender)
print(message)
```

```python
messages = []
for line in lines:
  if line.strip() == '':
    continue

  if ' - ' not in line:
    continue

  timestamp, remaining = line.split(' - ', 1)

  if ': ' not in remaining:
    continue

  sender, message = remaining.split(': ', 1)
  messages.append({
    'timestamp': timestamp,
    'sender': sender,
    'message': message
})
len(messages)
messages[0]
messages[-1]
messages[0]['sender']
messages[0]['message']


```

```python
time_gap = messages[1]['datetime'] - messages[0]['datetime']

print(time_gap)
print(time_gap.total_seconds() / 60)
```

```python
response_gaps = {}


for i in range(1, len(messages)):

    previous_sender = messages[i - 1]['sender']
    current_sender = messages[i]['sender']

    if previous_sender != current_sender:

        gap = messages[i]['datetime'] - messages[i - 1]['datetime']
        minutes = gap.total_seconds() / 60

        if current_sender not in response_gaps:
            response_gaps[current_sender] = []

        response_gaps[current_sender].append(minutes)


print(response_gaps.keys())
```

```python
average_response = {}


for sender in response_gaps:

    gaps = response_gaps[sender]

    total = sum(gaps)
    average = total / len(gaps)

    average_response[sender] = average


print(average_response)
```

```python
archetypes = [
    "THE SPAMMER",
    "THE GROUP MOM",
    "THE NIGHT OWL",
    "THE STORYTELLER",
    "THE DRAMA QUEEN",
    "THE GHOST",
    "THE COMEDIAN",
    "THE QUESTION MASTER"
]

print(archetypes)
```

```python
# Store consecutive-message burst lengths
burst_lengths = {}


for sender in participant_list:

    bursts = []
    current_burst = 0


    for msg in messages:

        if msg['sender'] == sender:
            current_burst += 1

        else:
            if current_burst > 0:
                bursts.append(current_burst)
                current_burst = 0



    # Save the final burst if the chat ended with this sender
    if current_burst > 0:
        bursts.append(current_burst)

    burst_lengths[sender] = bursts



print(burst_lengths)
```

```python
average_burst = {}


for sender in participant_list:

    bursts = burst_lengths[sender]

    total = sum(bursts)
    average = total / len(bursts)

    average_burst[sender] = average



print(average_burst)
```

```python
spammer_score = {}


for sender in participant_list:

    if average_burst[sender] > 3:
        spammer_score[sender] = 1
    else:
        spammer_score[sender] = 0


print(spammer_score)
```

```python
caring_keywords = [
    'okay',
    'safe',
    'eat',
    'sleep',
    'take care',
    'are you',
    'please',
    'reminder',
    'drink water',
    "don't forget"
]

caring_counts = {}

for sender in participant_list:

    count = 0

    for msg in messages:

        if msg['sender'] == sender:

            text = msg['message'].lower()

            for keyword in caring_keywords:

                if keyword in text:
                    count += 1


    caring_counts[sender] = count


print(caring_counts)
```

```python
group_mom_score = {}


for sender in participant_list:

    if caring_counts[sender] == max(caring_counts.values()):
        group_mom_score[sender] = 1
    else:
        group_mom_score[sender] = 0


print(group_mom_score)
```

```python
night_messages = {}
night_percentage = {}


for sender in participant_list:

    count = 0


    for msg in messages:

        if msg['sender'] == sender:

            hour = msg['datetime'].hour

            if hour >= 23 or hour <= 4:
                count += 1



    night_messages[sender] = count
    night_percentage[sender] = (count / message_counts[sender]) * 100



print(night_messages)
print(night_percentage)
```

```python
night_owl_score = {}


for sender in participant_list:

    if night_percentage[sender] > 60:
        night_owl_score[sender] = 1
    else:
        night_owl_score[sender] = 0


print(night_owl_score)
```

```python
word_counts_per_person = {}
average_words = {}

for sender in participant_list:

    total_words = 0
    total_messages = 0


    for msg in messages:

        if msg['sender'] == sender:

            words = msg['message'].split()

            total_words += len(words)
            total_messages += 1



    word_counts_per_person[sender] = total_words
    average_words[sender] = total_words / total_messages



print(average_words)
```

```python
storyteller_score = {}

for sender in participant_list:

    if average_words[sender] > 30:
        storyteller_score[sender] = 1
    else:
        storyteller_score[sender] = 0



print(storyteller_score)
```

```python
drama_counts = {}
drama_percentage = {}


for sender in participant_list:

    count = 0


    for msg in messages:

        if msg['sender'] == sender:

            text = msg['message'].strip()

            uppercase_condition = (
                len(text) >= 3 and text.isupper()
            )

            exclamation_condition = text.count('!') >= 2

            if uppercase_condition or exclamation_condition:
                count += 1



    drama_counts[sender] = count
    drama_percentage[sender] = (
        count / message_counts[sender]
    ) * 100



print(drama_counts)
print(drama_percentage)
```

```python
drama_queen_score = {}


for sender in participant_list:

    if drama_percentage[sender] > 30:
        drama_queen_score[sender] = 1
    else:
        drama_queen_score[sender] = 0



print(drama_queen_score)
```

```python
ghost_score = {}


for sender in participant_list:

    silent_percentage = (
        silent_days_count[sender] / len(all_dates)
    ) * 100

    if silent_percentage > 60:
        ghost_score[sender] = 1
    else:
        ghost_score[sender] = 0


print(ghost_score)
```

```python
comedy_words = ['lol', 'lmao', 'haha', 'rofl', 'lmfao']

comedy_counts = {}
comedy_percentage = {}


for sender in participant_list:
    count = 0

    for msg in messages:
        if msg['sender'] == sender:
            text = msg['message'].lower()

            for word in comedy_words:
                if word in text:
                    count += 1
                    break

    comedy_counts[sender] = count

    comedy_percentage[sender] = (
        count / message_counts[sender]
    ) * 100



print(comedy_counts)
print(comedy_percentage)
```

```python
comedian_score = {}


for sender in participant_list:

    if comedy_percentage[sender] == max(comedy_percentage.values()):
        comedian_score[sender] = 1
    else:
        comedian_score[sender] = 0


print(comedian_score)
```

```python
question_counts = {}
question_percentage = {}


for sender in participant_list:
    count = 0

    for msg in messages:
        if msg['sender'] == sender:
            text = msg['message'].strip()

            if text.endswith('?'):
                count += 1

    question_counts[sender] = count

    question_percentage[sender] = (
        count / message_counts[sender]
    ) * 100



print(question_counts)
print(question_percentage)
```

```python
question_master_score = {}


for sender in participant_list:

    if question_percentage[sender] > 25:
        question_master_score[sender] = 1
    else:
        question_master_score[sender] = 0



print(question_master_score)
```

```python
all_scores = {}


for sender in participant_list:

    all_scores[sender] = {
        "THE SPAMMER": spammer_score[sender],
        "THE GROUP MOM": group_mom_score[sender],
        "THE NIGHT OWL": night_owl_score[sender],
        "THE STORYTELLER": storyteller_score[sender],
        "THE DRAMA QUEEN": drama_queen_score[sender],
        "THE GHOST": ghost_score[sender],
        "THE COMEDIAN": comedian_score[sender],
        "THE QUESTION MASTER": question_master_score[sender]
    }



print(all_scores)
```

```python
final_archetype = {}


for sender in participant_list:

    scores = all_scores[sender]

    highest_score = max(scores.values())

    for archetype in archetypes:
        if scores[archetype] == highest_score:
            final_archetype[sender] = archetype
            break


print(final_archetype)
```
